```python
import urllib.request
from urllib.parse import quote  # 한글 파일명 URL 인코딩 처리
from langchain_community.document_loaders import (
    PyPDFLoader,
    PyMuPDFLoader,
    PDFPlumberLoader
)

# 실습용 PDF 다운로드 (2023 북한인권보고서)
pdf_url = quote(
    "https://raw.githubusercontent.com/ai-agent-kr/agent-tutorial/main/Ch02/2023_북한인권보고서.pdf",
    safe=":/"
)
urllib.request.urlretrieve(pdf_url, filename="2023_북한인권보고서.pdf")
```

```python
%%time
loader = PyPDFLoader("2023_북한인권보고서.pdf")
pages = loader.load_and_split()
print("청크의 수:", len(pages))
pages[3]
```

```python
%%time
loader = PyMuPDFLoader("2023_북한인권보고서.pdf")
pages = loader.load_and_split()
print("청크의 수:", len(pages))
```

```python
%%time
loader = PDFPlumberLoader("2023_북한인권보고서.pdf")
pages = loader.load_and_split()
print("청크의 수:", len(pages))
pages[3]
```
