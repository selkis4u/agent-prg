```python
import urllib.request
import re
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai import OpenAIEmbeddings

# 1. 파일 다운로드 및 읽기
urllib.request.urlretrieve(
    "https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt",
    filename="2016-10-20.txt"
)

with open("2016-10-20.txt", encoding="utf-8") as f:
    file = f.read()

print('텍스트 길이:', len(file))
```

```python
with open("2016-10-20.txt", encoding="utf-8") as f:
    full_text = f.read()

# 앞부분만  슬라이싱하여 sample_text 생성
sample_text = full_text[:150000]

print("sample_text 길이:", len(sample_text))
print("미리보기:\n", sample_text[:200])
```

```python
korean_sentence_regex = r"(?<=[.?!])\s+|\n+"

# 백분위수(Percentile) 방식 초기화
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=95,
    sentence_split_regex=korean_sentence_regex
)

# 문서 분할 수행 및 청크 개수 확인
texts = text_splitter.create_documents([sample_text])
print(f"분할된 청크의 수: {len(texts)}")  # 예제 결과: 9개
```

```python
korean_sentence_regex = r"(?<=[.?!])\s+|\n+"

# 표준편차(Standard Deviation) 방식 초기화
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=3,
    sentence_split_regex=korean_sentence_regex
)

# 문서 분할 수행 및 청크 개수 확인
texts = text_splitter.create_documents([sample_text])
print(f"분할된 청크의 수: {len(texts)}")  # 예제 결과: 3개 (백분위수 대비 청크 수 감소)
```

```python
korean_sentence_regex = r"(?<=[.?!])\s+|\n+"

# 사분위수(Interquartile) 방식 초기화
text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="interquartile",
    breakpoint_threshold_amount=1.5,
    sentence_split_regex=korean_sentence_regex
)

# 문서 분할 수행 및 청크 개수 확인
texts = text_splitter.create_documents([sample_text])
print(f"분할된 청크의 수: {len(texts)}")  # 예제 결과: 7개
```
