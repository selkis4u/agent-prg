```python
import numpy as np
from numpy import dot
from numpy.linalg import norm

def cos_sim(A: np.ndarray, B: np.ndarray) -> float:
    """
    두 1차원 벡터 A와 B의 내적(dot product)을
    각 벡터의 L2 노름(Norm, 크기)의 곱으로 나누어 코사인 유사도를 연산한다.
    """
    return dot(A, B) / (norm(A) * norm(B))

# 테스트 벡터 정의
vec1 = np.array([0, 1, 1, 1])
vec2 = np.array([1, 0, 1, 1])
vec3 = np.array([2, 0, 2, 2]) # vec2와 방향이 완전히 동일하고 크기만 2배 증가한 벡터

# 유사도 계산 및 출력
sim_1_2 = cos_sim(vec1, vec2)
sim_1_3 = cos_sim(vec1, vec3)
sim_2_3 = cos_sim(vec2, vec3)

print(f"벡터1과 벡터2의 유사도: {sim_1_2}")
print(f"벡터1과 벡터3의 유사도: {sim_1_3}")
print(f"벡터2와 벡터3의 유사도: {sim_2_3}")
```
